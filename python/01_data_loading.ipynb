```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("../data/tb_cust.csv")
transactions = pd.read_csv("../data/tb_transaction.csv")
products = pd.read_csv("../data/tb_transaction_product.csv")
fees = pd.read_csv("../data/tb_transaction_fees.csv")
```

```python
print("Customers:", customers.shape)
print("Transactions:", transactions.shape)
print("Products:", products.shape)
print("Fees:", fees.shape)
```

```python
display(customers.head())
display(transactions.head())
display(products.head())
display(fees.head())
```

```python
print("=== CUSTOMERS ===")
customers.info()

print("\n=== TRANSACTIONS ===")
transactions.info()

print("\n=== PRODUCTS ===")
products.info()

print("\n=== FEES ===")
fees.info()
```

```python
print("=== MISSING VALUES ===")

print("\nCustomers:")
print(customers.isnull().sum())

print("\nTransactions:")
print(transactions.isnull().sum())

print("\nProducts:")
print(products.isnull().sum())

print("\nFees:")
print(fees.isnull().sum())
```

```python
print("=== DUPLICATES ===")

print("Customers:", customers.duplicated().sum())
print("Transactions:", transactions.duplicated().sum())
print("Products:", products.duplicated().sum())
print("Fees:", fees.duplicated().sum())
```
